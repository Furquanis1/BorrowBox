## Locked admission rules

There is no `OPEN_FOR_ALL` community.

Supported admission modes:

```text
MANAGER_APPROVAL
LOCATION_VERIFIED
```

`MANAGER_APPROVAL` is the default for a newly created community.

V2.1.1 may persist the admission mode, but detailed join flows are implemented in V2.1.2.

## Creator rule

When an authenticated user creates a Community:

```text
Community.created_by = authenticated user
creator membership = ACTIVE
creator role = MANAGER
```

Never trust a client-supplied creator identity.

## Membership

Use the exact fields defined by the V2.1 schema, including:

```text
context_metadata JSON
```

Use Hibernate's native JSON mapping and prove it with a MySQL integration test.

## Community active-name uniqueness

Enforce:

```text
same creator + same normalized ACTIVE name = forbidden
different creators + same name = allowed
archived duplicates = allowed
```

The database must participate in the enforcement.

## Schema and seed

Use:

```text
schema.sql
ddl-auto=validate
ApplicationRunner
borrowbox.seed.enabled=true
```

Seed logic must be deterministic and idempotent.
