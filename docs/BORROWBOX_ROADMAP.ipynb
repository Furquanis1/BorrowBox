# BorrowBox V2 — Roadmap

**Status:** Living roadmap. Versions are milestones, not deadlines.

## v1.0.0 — Technical MVP ✅

Already released.

## v2.1 — Community + Ownership Foundation

**Goal:** Make BorrowBox genuinely community-based.

Scope:

- Community creation
- Membership
- Manager
- Member roles/context
- Community type
- Community location/boundary concept
- Join/leave
- Verification concept
- Community rules
- Owned assets
- Asset units
- Community listings
- Selective community visibility
- Community-scoped Explore
- Community-scoped inventory

**Completion test:** A user can create/join a real community, operate under a community role, own an asset once, expose it to selected communities, and have all communities reflect the same underlying physical availability.

## v2.2 — Transaction Engine

**Goal:** Complete borrowing without relying on outside tools.

Scope:

- Structured requests
- Purpose/duration
- Approval/rejection
- Counter-offers
- Transaction messaging
- Pickup coordination
- Representative handling
- Four live evidence moments
- Lender handover confirmation
- Borrower dispute window
- Loan timer
- Notifications
- Extensions
- Return workflow
- Overdue handling
- Race-condition-safe reservation/queueing

**Completion test:** A borrower can request, get approved, coordinate pickup, complete a verified handover, borrow, receive reminders, extend where appropriate, return with evidence, and close the transaction.

This milestone may move earlier as useful.

## v3.x — Advanced Resolution

Scope:

- AI-assisted condition comparison
- AI anomaly flags
- Borrower appeal
- Blind tribunal
- Anonymous peer review
- Consensus resolution
- Reputation consequences

## Roadmap principles

- Build in dependency order.
- Prefer complete vertical slices.
- Do not introduce AI before the evidence and transaction foundations are mature.
- Every milestone must be manually demonstrable.
- New features must support the core product philosophy.
