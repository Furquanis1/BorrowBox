# BorrowBox V2 — Roadmap

**Status:** Living roadmap. Versions are milestones, not deadlines.

## v1.0.0 — Technical MVP ✅

Already released.

## V2 database strategy

V2.1 starts from a **fresh database**.

The V2 implementation does not migrate the V1 development/demo dataset. V1 remains preserved as the released v1 product, while V2 uses its own clean schema and deterministic seed data.

A repeatable initialization/reset path is part of the V2 development workflow.

## v2.1 — Community + Ownership Foundation

**Goal:** Make BorrowBox genuinely community-based while establishing the global ownership model.

### V2.1.1 — Community + Membership foundation
- Community entity
- Membership entity
- Roles/status/verification fields
- `context_metadata` JSON
- Creator becomes manager
- Community APIs
- Membership APIs
- Minimal frontend community shell / active-community model
- MySQL schema + deterministic seed foundation
- Backend + frontend verification tests

### V2.1.2 — Community admission
- Manager-approved join flow
- Location-verified instant join
- Leave/rejoin lifecycle
- No open-for-all communities
- Server-side verification
- Explicit, non-continuous location checks

### V2.1.3 — Community rules foundation
- Basic community rule storage
- Manager-controlled rule management
- Membership-policy integration points
- No transaction/borrowing rules yet

### V2.1.4 — Asset + AssetUnit ownership
- Global Asset ownership
- Immediate AssetUnit materialization
- Atomic Asset + N units creation
- Aggregate inventory model

### V2.1.5 — CommunityListing
- Selective community visibility
- Owner ACTIVE-membership authorization
- One Asset across multiple Communities
- Shared physical inventory pool

### V2.1.6 — Community Explore + Inventory
- Community-scoped Explore
- Community-scoped owner inventory
- Aggregate availability
- Public hiding of AssetUnit IDs
- Cross-community availability tests

### V2.1.7 — V2.1 seed/reset + final QA
- Deterministic complete V2.1 seed
- Idempotent re-run
- Clean database reset
- Backend/frontend/E2E tests
- Docker verification
- Manual acceptance flow

**V2.1 completion test:** A user can create/join a bounded community with an attributed membership, own an asset once, selectively expose it to multiple communities, and see one shared physical inventory state reflected consistently across those communities.

## v2.2 — Transaction Engine

**Goal:** Complete borrowing without relying on outside tools.

Scope:

- Structured requests
- Purpose/duration
- Approval/rejection
- Counter-offers
- Transaction messaging
- Pickup coordination
- Representative handling
- Four live evidence moments
- Lender handover confirmation
- Borrower dispute window
- Loan timer
- Notifications
- Extensions
- Return workflow
- Overdue handling
- Race-condition-safe reservation/queueing

**Completion test:** A borrower can request, get approved, coordinate pickup, complete a verified handover, borrow, receive reminders, extend where appropriate, return with evidence, and close the transaction.

## Roadmap principles

- Build in dependency order.
- Prefer complete vertical slices.
- Do not introduce AI before the evidence and transaction foundations are mature.
- Every milestone must be manually demonstrable.
- New features must support the core product philosophy.
