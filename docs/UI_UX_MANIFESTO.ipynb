# BorrowBox UI/UX Manifesto

This notebook locks in the frontend design constraints for the BorrowBox adaptive UI. These rules are authoritative for the desktop-first correction and the Phase 1-4 roadmap. No backend files, schemas, or APIs may be changed as part of this work.


## Typography

- **Family:** Instrument Sans for all UI text, including headings and `.serif` elements.
- **Weights:** 400 (body), 500 (labels), 600 (controls/tabs), 700 (headings/emphasis).
- **Load path:** Google Fonts via `frontend/index.html`; headings are forced to Instrument Sans with `font-family: 'Instrument Sans', sans-serif !important;` to eradicate the old serif face.
- **Rule:** No DM Sans, DM Serif Display, or other legacy display fallback remains in the UI.


## Colors

| Role | Value | Usage |
| --- | --- | --- |
| Canvas | `#f5f6f7` | App background behind the centered container |
| Cards / surfaces | `#ffffff` | Cards, tab pills, header surfaces |
| Ink / brand | `#212529` | Strong text, active pills, floating nav, primary buttons |

The palette is a stark charcoal system. Teal tokens are remapped to neutral charcoal values so legacy accents never reintroduce a green cast.


## Layout

- **Desktop:** `.mobile-app-container` is centered with `max-width: 1400px`, `margin: 0 auto`, and `padding: 40px 48px`. The background is transparent and the old mobile shadow is removed.
- **Desktop nav:** at `min-width: 768px`, `.floating-nav` docks near the top (`top: 32px`, centered, `width: fit-content`, `gap: 32px`) and the container receives `padding-top: 120px`.
- **Mobile thumb zone:** at `max-width: 767px`, `.floating-nav` returns to the bottom (`bottom: 24px`, centered, `width: calc(100% - 48px)`, space-between) so primary actions stay within thumb reach.
- **Tabs:** `.community-tabs` are horizontally scrollable with pill-shaped links; the active pill uses `[aria-current="page"]` with charcoal background and white text.


## Strict Phase 1-4 Roadmap

1. **Phase 1 - Mobile-first shell & design system:** Instrument Sans, charcoal tokens, responsive container, floating nav, and simplified header.
2. **Phase 2 - Component rollout:** Reuse the locked tokens for cards, buttons, tabs, and drawers. Preserve existing behavior while removing legacy desktop chrome.
3. **Phase 3 - Visual QA & accessibility:** Verify 1400px desktop cap, mobile thumb zone, focus states, reduced motion, and contrast against the charcoal palette.
4. **Phase 4 - Test alignment:** Update frontend selectors and Cypress expectations only; never change backend behavior to satisfy UI tests.

**Non-negotiable:** No backend files, database changes, or API contract changes in any phase. All work stays inside `frontend/` plus this design document.
