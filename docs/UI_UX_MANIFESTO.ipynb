# BorrowBox UI/UX Manifesto

This notebook is the authoritative desktop-first design contract for BorrowBox (Option B). Its visual language has evolved into a TripGlide-inspired airy system: generous padding, soft tonal overlays, 20px–24px rounded corners, and 40px floating/docked bottom-right round action buttons, all within the `#212529` / `#f5f6f7` / `#ffffff` palette. Backend files, API contracts, database schemas, and React Router route definitions remain out of scope for this work.

## Typography

- **Family:** Instrument Sans universally — headings, body, labels, controls, tabs, and any legacy `.serif` elements all use the same humanist sans.
- **Ink:** `#212529` charcoal is the universal text color; muted text uses restrained neutral steps, never a second typeface.
- **Weights:** 400 (body), 500 (labels), 600 (controls/tabs), 700 (headings/emphasis).
- **Load path:** Google Fonts via `frontend/index.html`; DM Sans, DM Serif Display, and other legacy display faces are removed from the UI.


## Palette & Surfaces

| Role | Value | Usage |
| --- | --- | --- |
| Canvas | `#f5f6f7` | Warm off-white application canvas behind the centered main column |
| Surface | `#ffffff` | Cards, rails, drawers, modals, and raised controls |
| Rail surface | `#ffffff` | Airy, borderless fixed left visual rail |
| Ink | `#212529` | Charcoal text, active indicators, and primary actions |
| Soft overlay | `#f5f6f7` | Gentle active pills, hover wells, and tonal layering |
| Sage wash | `#EDEADE` | Restrained community and identity emphasis |
| Hairline | `1px solid #e5e7eb` | Retained only where structure still needs it; replaced by tonal layering on the rail and cards |
| Card lift | `0 4px 16px rgba(0,0,0,0.03)` | Barely-there elevation for compact interactive cards |

Depth comes from whitespace, soft tonal overlays, and restrained elevation rather than heavy borders or large shadows. Shadows are reserved for transient overlays, drawers, modals, and toasts.

## Layout Skeleton

- **Desktop (`>= 768px`):** `.desktop-shell` is a flex row. `.desktop-side-rail` is a fixed `240px` white, borderless visual rail with `24px 18px` padding, an editorial brand header, primary global navigation, community links, and a circular profile photo with a small notification badge at the bottom.
- **Rail interactions:** menu items are ample-height links with `16px` border-radius; the active item uses a gentle `#f5f6f7` pill background instead of a heavy box.
- **Main canvas:** `.desktop-main-area` fills the remaining width with a `240px` left offset, `min-height: 100vh`, `#f5f6f7` background, and `40px 48px` padding. `.desktop-canvas` is a centered `max-width: 1040px` flex column with a `24px` gap.
- **Search & filters:** the search surface is a borderless `48px` white pill with `24px` radius and a barely-there shadow. Category filters are horizontal `20px` pills; the active filter inverts to `#212529`.
- **Community tabs:** `.community-tabs` are underlined text links with no boxy borders or container edge; the active tab uses a bold `#212529` underline.
- **Natural document flow:** the window is the only scroll container for main page content. There are no `100vh` page scroll traps, inner canvas scrollers, or full-width floating bottom navigation bars. Small round action buttons may be docked inside cards and overlays.
- **Narrow viewports (`< 768px`):** the rail collapses into a compact top bar with icon-only primary navigation so small screens never break.

## TripGlide-Inspired Airy Redesign

- **Spacing:** generous `40px 48px` canvas padding and a `24px` content stack create a calm, editorial desktop rhythm.
- **Soft tonal overlays:** `#f5f6f7` active pills and hover wells, `#EDEADE` quiet emphasis, and `#ffffff` surfaces replace heavy outlines.
- **Radii:** `20px` immersive cards, `24px` drawers and modals, `20px` filter and schedule-date pills, `16px` rail links, and `8px` standard controls.
- **Round actions:** compact cards use a `36px` bottom-right `.btn-circular` arrow; larger surfaces use `40px` floating/docked bottom-right round action buttons. Wide “Borrow” buttons are replaced by restrained right-facing chevrons/arrows.
- **Cards:** compact interactive cards use `160px` media, no visible outer hairline, tiny floating “Pre-loved” and availability labels, and a subtle `translateY(-4px)` hover.
- **Drawers & modals:** `#ffffff` surfaces with `24px` radii, generous padding, pill-styled schedule dates, standard search/dropdown/input structures, and a dark `#212529` pill CTA labeled “Request Borrow”.
- **Continuity:** all card callbacks, custom behaviors, backend contracts, and routing remain untouched; this is a presentation-system refinement only.

## Part 1-4 Roadmap

1. **Part 1 (Desktop Shell & Left Rail):** Establish the fixed `240px` rail, fluid centered canvas, natural document flow, and underlined community tabs while preserving existing behavior and routes.
2. **Part 2 (Airy Compact Item Cards):** Rebuild item cards on a three-column desktop grid using white surfaces, soft elevation, `20px` corners, floating status labels, and bottom-right circular arrow actions.
3. **Part 3 (Drawers & Secondary Views):** Align drawers, dialogs, and secondary pages with the same typography, palette, `24px` radii, and pill controls without changing backend contracts.
4. **Part 4 (Cypress Alignment & Clean Git Squash):** Update frontend selectors and Cypress expectations only, then produce a clean, reviewable Git history.

**Non-negotiable:** No backend files, database changes, API contract changes, or React Router route changes in any part. All work stays inside `frontend/` plus this design document.
