# BorrowBox V1 → V2.1 Migration Analysis — Historical Reference

**Status:** Historical reasoning only — NOT an implementation specification.

V2.1 starts from a fresh database and deterministic seed data.

The V1 database is preserved in Git/history but is not imported by the V2.1 runtime.

This document records why the V2 ownership model was redesigned around:

```text
User
Membership
Community

Asset
AssetUnit
CommunityListing
```

It must not be used to generate a V1→V2 migration layer.

Do not:

```text
read V1 tables at runtime
convert legacy Groups into V2 Communities
convert legacy Items into V2 Assets
convert legacy BorrowRequests into V2 Transactions
```

unless a future product decision explicitly reintroduces data import.
