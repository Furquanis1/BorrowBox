# BorrowBox V2 — Transaction Specification

**Status:** Core business specification.

## Definition

A transaction represents one complete physical lending event involving:

- One lender
- One borrower
- One community
- One reserved physical asset unit (or unit from an asset pool)

## State machine

```text
REQUESTED
   ↓
APPROVED
   ↓
AWAITING_HANDOVER
   ↓
HANDOVER_REPORTED
   ↓
ACTIVE
   ↓
RETURN_INITIATED
   ↓
RETURN_HANDOVER_REPORTED
   ↓
COMPLETED
```

Derived conditions:

```text
ACTIVE + approaching due time = DUE_SOON
ACTIVE + current time > due time = OVERDUE
```

Exceptions may include:

```text
CANCELLED
REJECTED
HANDOVER_DISPUTED
RETURN_DISPUTED
FLAGGED
EXPIRED
```

## Actor permissions

### Borrower

- Create/cancel eligible requests
- Specify purpose/duration
- Coordinate pickup
- Confirm/dispute handover
- Request extension
- Initiate return
- Capture return evidence
- Report problems

### Lender

- Approve/reject
- Counter-offer
- Coordinate pickup
- Capture borrow-side evidence
- Report handover
- Confirm return
- Report problems

### System

- Start authoritative timing from handover confirmation
- Send reminders
- Calculate due/overdue conditions
- Generate notifications
- Trigger configured escalations

### Community manager

- Review membership
- Review flags/incidents
- Apply community-defined moderation actions

## Borrow handover

Lender captures:

1. Live pre-handover photo
2. Live handover photo showing the item and relevant recipient/participant

Lender confirms handover.

At that moment:

```text
Loan accountability clock starts
+
Borrower dispute/confirmation window opens
```

The envisioned default window is 30 minutes.

Borrower may:

- Confirm receipt
- Report a problem
- Dispute non-receipt

A dispute should not automatically determine guilt.

## Return handover

Borrower captures:

1. Live pre-return photo
2. Live return-handover photo showing item + relevant receiving participant

Borrower submits the return.

Lender confirms receipt.

Only then does the transaction complete.

## Representatives

If a friend/colleague/other participant physically performs a handover:

```text
Account owner
Physical participant
Acting on behalf of
Relationship/role
```

must be recorded.

The transaction conversation should make the arrangement visible to the relevant parties.

## Extension

An extension is an event, not a destructive update.

The history should retain:

```text
Original due date
Request
Requested new date
Decision
Final due date
```

## Overdue

Overdue should derive from authoritative time.

Community rules may define a grace period before escalation.

## Evidence

Four evidence moments:

```text
1. Lender — before lending
2. Lender — actual borrow handover
3. Borrower — before returning
4. Borrower — actual return handover
```

Evidence is transaction-scoped and should not become public profile content.

## Disputes

Examples:

- Non-receipt
- Pre-existing damage
- Wrong item
- Missing component
- Return disagreement

AI inspection and a blind tribunal are future escalation mechanisms.
