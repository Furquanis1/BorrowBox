# BorrowBox V2 — Product Specification

**Status:** Living functional specification.

## Member model

The primary actor is a BorrowBox member.

The same member may act as:

- Borrower in one transaction
- Lender in another transaction
- Community manager in a community where they hold the manager role

Borrower/lender are transaction roles, not permanent user types.

## Community capabilities

A member should eventually be able to:

- Discover communities
- Create a community
- Join a community
- Leave a community
- Switch active community
- View members
- See community type and rules
- Declare their role/context inside the community

A manager should be able to:

- Review membership
- Approve/reject members
- Ask for clarification
- Remove/suspend members
- Manage rules
- Manage community location/boundary

## Asset capabilities

A user owns an asset once.

They can choose which communities may see that asset.

Example:

```text
Ahmed owns:
  Football × 2

Listed in:
  Community A
  Community B
  Community C
```

This is one ownership pool, not three independent inventory records.

## Shared quantity

Availability is tied to physical quantity.

```text
Total = 2
Borrowed = 0
Available = 2
```

After one physical unit is borrowed:

```text
Total = 2
Borrowed = 1
Available = 1
```

All selected community listings reflect the same underlying availability.

## Transaction flow

### Request

Borrower chooses an available asset/listing and provides:

- Purpose
- Requested duration
- Optional message

### Approval

Lender may:

- Approve
- Reject
- Counter-offer

### Pickup coordination

Transaction messaging is used to coordinate:

- Who picks up
- Who hands over
- Representative involvement
- Time/place
- Notes

### Borrow handover

Lender captures:

1. Live pre-handover photo
2. Live transfer photo including item + relevant recipient/participant

Lender then confirms handover.

The accountability clock starts and a short borrower dispute window opens.

### Active borrowing

The system tracks:

- Start time
- Due time
- Remaining time
- Notifications
- Overdue state

### Extension

Borrower can request a new due date inside the transaction conversation.

Lender can accept, reject, or counter.

The transaction preserves the original and revised due dates.

### Return

Borrower:

1. Initiates return.
2. Takes live pre-return photo.
3. Hands the item to the lender/representative.
4. Takes live return-handover photo.
5. Submits the return.

Lender confirms receipt.

The transaction then completes.

## Problems and flags

Possible issues include:

- Item not received
- Wrong item
- Existing damage
- Missing accessory
- Handover problem
- Late/non-return
- Other community-defined issues

A flag is an incident record; it is not automatically a punishment.

## Reputation

A member profile should eventually expose:

- Items lent
- Items borrowed
- On-time return rate
- Successful transactions
- Reputation/trust
- Community badges
- Contribution recognition

Reputation should be derived from recorded behavior rather than arbitrary activity volume.
