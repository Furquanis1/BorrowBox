# BorrowBox V2 — Architecture & Product Decisions

**Status:** Append-only decision log.

## ADR-001 — User is the fundamental actor

**Decision:** A BorrowBox user is a member. Borrower/lender are contextual transaction roles.

**Reason:** The same person can lend in one transaction and borrow in another.

## ADR-002 — Communities are first-class product boundaries

**Decision:** Communities represent real-world/social contexts with type, location, manager, membership, and rules.

**Reason:** Trust is grounded in the surrounding community.

## ADR-003 — Membership is a first-class entity

**Decision:** Community-specific role/context belongs to Membership.

**Reason:** One person can have different roles and attributes in different communities.

## ADR-004 — Assets are globally owned; listings are community-specific

**Decision:** One owned asset pool may be listed in multiple communities.

**Reason:** Prevents duplicate physical inventory and keeps availability synchronized.

## ADR-005 — Availability is backend authoritative

**Decision:** The backend/database owns the truth for quantity and availability.

**Reason:** Concurrent requests cannot safely be resolved by frontend state.

## ADR-006 — Server determines request order

**Decision:** Client timestamps do not establish queue position.

**Reason:** Device clocks differ and are not authoritative.

## ADR-007 — Physical handover is a first-class transaction event

**Decision:** Handover is explicitly recorded.

**Reason:** Physical responsibility changes need durable evidence.

## ADR-008 — Four evidence moments

**Decision:** A normal transaction plans four live-photo evidence moments.

**Reason:** Two establish borrow-side condition/transfer and two establish return-side condition/transfer.

## ADR-009 — Lender handover starts accountability

**Decision:** Lender confirmation starts the borrowing accountability clock and opens a short borrower dispute window.

**Reason:** The lender should not remain indefinitely exposed while waiting for a borrower acknowledgment.

## ADR-010 — Evidence is not absolute truth

**Decision:** Photos, messages, timestamps, and confirmations are evidence, not infallible proof.

**Reason:** Legitimate disputes and ambiguous evidence remain possible.

## ADR-011 — Privacy by design

**Decision:** Evidence photos are transaction records, not public profile/feed content.

**Reason:** They can contain faces and should be visible only to appropriate participants/moderators.

## ADR-012 — No continuous location surveillance

**Decision:** Location can support community verification but should not imply permanent tracking.

**Reason:** BorrowBox is a trust/community system, not a movement-monitoring service.

## ADR-013 — AI and tribunal are deferred

**Decision:** AI inspection and blind tribunal functionality are later-stage capabilities.

**Reason:** They require mature community, transaction, evidence, and reputation foundations.
