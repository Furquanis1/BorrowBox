# BorrowBox V2 — Vision

**Status:** Living product vision.

## Core idea

BorrowBox is a private sharing network for small, real-world communities where members can discover, lend, borrow, and track physical items through a structured and trustworthy transaction lifecycle.

A person is not permanently a borrower or lender. They are a member who may need something today and lend something tomorrow.

> **I need this, but I do not need to own this.**

## Core philosophy

- BorrowBox is not a buying/selling or rental marketplace.
- Small, bounded communities create the foundation of trust.
- A member's status comes from responsible participation and useful contribution.
- BorrowBox reduces the social friction of asking for help.
- The system acts as a timekeeper and digital transaction ledger.
- Important physical events should be recorded rather than left to memory or informal conversation.
- Reliability, speed, clarity, privacy, and UX are product requirements.
- Location may verify community membership, but continuous movement tracking is not the goal.
- AI inspection and a blind tribunal are later-stage capabilities, not V2 starting requirements.

## Social loop

```text
Member
  ↓
Needs something / Has something
  ↓
Borrow / Lend
  ↓
Successful transaction
  ↓
Contribution + reliability recorded
  ↓
Recognition / reputation
  ↓
More trust
  ↓
More useful participation
  ↺
```

## Community philosophy

A community represents a real-world or tightly bounded social context: hostel, college department, office, housing society, club, friend circle, or similar group.

Each community has:

- Type
- Physical location/boundary
- Manager
- Membership policy
- Member roles/context
- Community rules

Membership is controlled. Communities are not intended to be open anonymous public pools.

## Trust and physical handover

A transaction plans four live-photo evidence moments:

1. Lender takes a live photo immediately before lending.
2. Lender takes a live handover photo showing the item and relevant recipient/participant.
3. Borrower takes a live photo immediately before returning.
4. Borrower takes a live return-handover photo showing the item and relevant receiving participant.

These photos establish condition and transfer evidence.

A physical representative may act for a lender or borrower when appropriate; the transaction must record who actually performed the handover and on whose behalf.

## Handover protection

When the lender confirms the item has been handed over:

- The borrowing accountability clock starts.
- The borrower receives a short confirmation/dispute window.
- The borrower can confirm receipt, report a problem, or dispute non-receipt.
- If no dispute is raised before the window closes, the handover can settle according to community rules.

The exact window is a product/configuration decision; a 30-minute default is currently envisioned.

## Long-term vision

The mature BorrowBox vision grows in layers:

```text
Community
  ↓
Transactions
  ↓
Evidence
  ↓
Reputation
  ↓
Flags / moderation
  ↓
Condition intelligence
  ↓
AI inspection / appeals / blind tribunal
```

## Non-goals

- Buying and selling
- Rental-marketplace behavior
- Generic public social networking
- Continuous member location surveillance
- Jumping directly to AI before the transaction and trust foundations exist
