# BorrowBox V2 — Architecture

**Status:** Living technical blueprint.

## Core relationship model

```text
User
 │
 ├── Membership ───────► Community
 │                          │
 │                          ├── Rules
 │                          ├── Members
 │                          ├── Listings
 │                          ├── Transactions
 │                          └── Flags
 │
 └── Asset
      │
      ├── AssetUnit
      │      └── Transaction
      │
      └── CommunityListing
```

## User

Global authentication identity.

A User should not be permanently classified as borrower/lender/student/employee/etc.

## Community

Represents a bounded real-world/social group.

Important concepts:

- Type
- Location/boundary
- Manager
- Membership
- Rules
- Shared listings

## Membership

First-class relationship between User and Community.

Potential data:

- user
- community
- role
- status
- verification method/state
- joined/verified timestamps
- community-specific context

Community-specific context could include fields such as hostel room/floor, office department, or college year/section without making those fields global user attributes.

## Asset

Represents the owner's actual asset pool.

Example:

```text
Asset
  owner = Ahmed
  title = Football
  totalQuantity = 2
```

## AssetUnit

Represents one individually trackable physical unit when quantity or condition tracking requires it.

This permits:

```text
Football
  ├── Unit 1
  └── Unit 2
```

while the UI can still present the collection as one asset.

## CommunityListing

Links an owned asset pool to a selected community.

This prevents duplicated "copies" of the same physical asset across communities.

## Transaction

The central business object.

It references:

- Asset/asset unit
- Community
- Lender
- Borrower
- Messages
- Handovers
- Evidence
- Notifications/events
- Flags
- Reputation events

## Authority boundaries

The backend/database is authoritative for:

- Physical quantity
- Availability
- Reservation
- Transaction state
- Due dates
- Request ordering
- Permissions
- Reputation events
- Membership status

The frontend displays and requests changes; it does not decide business truth.

## Concurrency

Client timestamps must not determine queue order.

The server/database establishes authoritative ordering.

Concurrent requests against scarce units must use a database-safe reservation/locking strategy so a physical unit cannot be allocated twice.

## Event history

Important physical and contractual events should be durable and reconstructable:

```text
Who
What
When
Which community
Which asset/unit
What agreement
What evidence
What confirmations
What extensions
What outcome
```

## Future extensibility

The architecture should leave room for:

- Notification channels
- Community-specific membership schemas
- Condition history
- Moderation
- AI inspection
- Appeals and tribunal review

These capabilities are intentionally incremental.


## V2 database boundary — LOCKED

V2.1 uses a **fresh database/schema**.

There is no V1-to-V2 migration layer in the V2 application baseline.

The V1 database and development data remain historical v1 material preserved in Git/release history. The V2 runtime starts from:

```text
fresh database
→ V2.1 schema
→ deterministic seed data
→ V2 application
```

The V2 schema must not depend on V1 runtime tables such as `borrow_groups`, `group_users`, or the V1 `items` table.

A repeatable seed/reset mechanism is part of the V2 development workflow.
