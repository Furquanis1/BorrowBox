## Hard boundaries

```text
V2.1 starts from a fresh database.
V1 data is not migrated.
No OPEN_FOR_ALL communities.
Community admission is manager approval or explicit location verification.
Community + Membership is the first implementation slice.
Assets/AssetUnits/Listings come after the community foundation.
Transactions come later in V2.2.
AI/tribunal come much later.
```
