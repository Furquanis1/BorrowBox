# BorrowBox V2 — Documentation Index

This notebook is the entry point to the living V2 planning set.

## Documents

1. **BORROWBOX_V2_VISION.ipynb** — Product identity, philosophy, and long-term direction.
2. **BORROWBOX_PRODUCT_SPEC.ipynb** — Functional behavior and user capabilities.
3. **BORROWBOX_ARCHITECTURE.ipynb** — Entities, relationships, authority boundaries, and technical principles.
4. **BORROWBOX_TRANSACTION_SPEC.ipynb** — Transaction lifecycle, evidence, timing, permissions, and disputes.
5. **BORROWBOX_ROADMAP.ipynb** — Ordered milestones and completion criteria.
6. **BORROWBOX_DECISIONS.ipynb** — Append-only record of important product and architectural choices.

## Planning hierarchy

```text
VISION
  ↓
PRODUCT SPECIFICATION
  ↓
ARCHITECTURE
  ↓
TRANSACTION SPECIFICATION
  ↓
ROADMAP
  ↓
IMPLEMENTATION

DECISIONS
  ↕
All layers
```

## Maintenance rules

- Vision changes only when a core product principle changes.
- Product Spec changes when intended user behavior changes.
- Architecture changes when the technical/domain model changes.
- Transaction Spec changes when states, permissions, timing, or evidence rules change.
- Roadmap changes as priorities and dependencies evolve.
- Decisions records why important choices were made.
